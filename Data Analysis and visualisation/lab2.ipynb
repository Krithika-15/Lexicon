## Task 3 : Average beds per room type

```python
listings["room_type"].value_counts()
```

```python
listings.groupby("room_type")["beds"].mean().round(2)
```

```python
# groupby puts the listings into groups by room type, then I take the mean of beds in each group.

# Note to self: count is NOT the number of beds!
# It's how many listings actually have a beds value filled in.
# Some hosts left beds empty (NaN), so those listings are skipped.
# That's why Entire home/apt shows 3680 here, but value_counts() said 3760.
# (A 0 would still be counted. Only empty ones are skipped.)

# The mean is just: all the beds added up / count
# sum of all beds in entire home/apt / 3680 (here among 3760 rows, only 3680 rows have bed value filled in)
# e.g. Entire home/apt: 9128 / 3680 = 2.48

listings.groupby("room_type")["beds"].agg(["count", "mean"])
```

**Answer:** I trust the **Hotel room** average (1.00) the least. It's based on only 3 listings, so it can't really say what hotel rooms are like in general. **Entire home/apt** is the most trustworthy one, because its average comes from 3680 listings.

## Task 4 : Reviews per neighbourhood

```python
listings["neighbourhood_cleansed"].value_counts()
```

```python
listings.groupby("neighbourhood_cleansed")["number_of_reviews"].agg(["count", "mean", "median"]).round(2)
```

```python
listings.groupby("neighbourhood_cleansed")["number_of_reviews"].agg(["count", "mean", "median"]).round(2).sort_values("median")
```

**Answer:** The lowest median is **Skärholmens (3)** and the highest is **Södermalms (17)**.

In Södermalms the mean (59.82) is much bigger than the median (17). Most listings there have only a few reviews, but a few have hundreds or even over a thousand. Those few pull the mean up, while the median stays in the middle, so the median shows a normal listing better.

## Task 6 : Two categories at once

```python
listings["host_is_superhost"].value_counts()
```

```python
# crosstab is value_counts done once per group, laid out as a grid

pd.crosstab(listings["neighbourhood_cleansed"], listings["host_is_superhost"])
```

```python
# eg : bromma has total 281, so 261/281 for f and 20/281 for t
# normalize="index" means per row. so each cell is divided by its own row's total
# normalize gives shares as decimals between 0 and 1 for eg: 0.071
# multiplying by 100 turns them into percent for eg : 7.1

(pd.crosstab(listings["neighbourhood_cleansed"], listings["host_is_superhost"], normalize='index')*100).round(1)
```

```python
# sorting based on column t to find the highest and lowest share of superhosts
((pd.crosstab(listings["neighbourhood_cleansed"], listings["host_is_superhost"], normalize='index')*100).round(1)).sort_values("t")
```

**Answer:** Södermalms has the highest share of superhosts (23.8 %) and Bromma has the lowest (7.1 %). In Bromma only 20 of 281 listings have a superhost, so about 7 in every 100.
